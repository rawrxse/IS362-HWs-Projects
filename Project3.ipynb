In this project, I use the Chinook database to find the tracks and albums purchased by each customer. I will join five SQL tables and display the results in a pandas DataFrame, sorted by the customers’ last and first names.


I used Python's built-in sqlite3 module to connect to SQLite, so no separate driver installation was required.

My connection string is `file:Project 3/Chinook_Sqlite.sqlite?mode=ro`. It points to the database inside the Project 3 folder and opens it in read-only mode.

sqlite3 documentation: https://docs.python.org/3/library/sqlite3.html

```python
import sqlite3
import pandas as pd

conn = sqlite3.connect(
    "file:Project 3/Chinook_Sqlite.sqlite?mode=ro",
    
    uri=True
)

pd.read_sql_query("SELECT name FROM sqlite_master WHERE type='table';", conn)
```

## Joining the tables

I connect customers to their invoices using CustomerId, then connect each invoice to its purchased items using InvoiceId. TrackId identifies the purchased track, and AlbumId connects that track to its album.

The query selects the customer's last and first names, track name, and album title. I sort the results by last name and then first name.



```python
query = """
SELECT Customer.LastName,
       Customer.FirstName,
       Track.Name,
       Album.Title
FROM Customer
JOIN Invoice
    ON Customer.CustomerId = Invoice.CustomerId
JOIN InvoiceLine
    ON Invoice.InvoiceId = InvoiceLine.InvoiceId
JOIN Track
    ON InvoiceLine.TrackId = Track.TrackId
JOIN Album
    ON Track.AlbumId = Album.AlbumId
ORDER BY Customer.LastName, Customer.FirstName;
"""

df = pd.read_sql_query(query, conn)

df.head()
```

The first five rows show tracks purchased by Roberto Almeida. Each row includes the customer's last and first names, the purchased track, and its album. The same customer appears multiple times because they purchased multiple tracks.
