
Project 1: Comparing Airline Arrival Delays

This project compares the arrival records of Alaska Airlines and America West across five destinations. I will use pandas to read the flight data from a CSV file, calculate delay percentages, and compare the airlines both overall and by destination. Comparing percentages will account for the fact that the airlines operated different numbers of flights.


Loading the Data

I will begin by importing pandas and reading the CSV file into a DataFrame. Displaying the DataFrame will allow me to confirm that all flight information was loaded correctly before beginning the analysis.


```python
import pandas as pd

airlines = pd.read_csv("Project1 Airlines(Sheet1).csv")
airlines

```


Calculating Delay Percentages by Destination

 I will first calculate the total number of flights for each airline and destination. I will then divide the delayed flights by the total flights and multiply by 100 to find the delay percentage. The number of delayed flights alone does not provide a fair comparison because the airlines did not have the same amount of flights. 

```python
airlines["Total Flights"] = airlines["On Time"] + airlines["Delayed"]

airlines["Delay Percentage"] = (
    airlines["Delayed"] / airlines["Total Flights"] * 100
).round(2)

airlines
```

Calculating the Overall Delay Percentage

Next, I will combine the flight counts from all five cities for each airline. After finding the total on-time, delayed, and overall flights, I will calculate each airline's overall delay percentage. This will show which airline had the lower delay rate overall
.


```python
overall_results = airlines.groupby("Airline")[
    ["On Time", "Delayed", "Total Flights"]
].sum()

overall_results["Delay Percentage"] = (
    overall_results["Delayed"] / overall_results["Total Flights"] * 100
).round(2)

overall_results
```

Comparing the Airlines by Destination

The overall results show that America West had a lower delay percentage than Alaska. I will also arrange the results by cities to see whether America West performed better in each individual city or whether the overall result was influenced by the number of flights each airline operated at different destinations.



```python
destination_results = airlines[
    ["Destination", "Airline", "Delay Percentage"]
].sort_values(by=["Destination", "Airline"])

destination_results
```

Conclusion


When all destinations are combined, Alaska had 501 delayed flights out of 3,775 total flights, giving it an overall delay rate of 13.27%. America West had 787 delayed flights out of 7,225 total flights, giving it a lower overall delay rate of 10.89%.

However, Alaska had a lower delay percentage at every individual destination.  America West operated most of its flights in Phoenix, where delay rates were relatively low, while many of Alaska's flights went to Seattle, where delay rates were higher. Therefore, America West appears better based on the combined percentage due to operating more flights, but Alaska performed better when each destination is compared separately.This shows why it is helpful to look beyond the overall percentage when comparing.
