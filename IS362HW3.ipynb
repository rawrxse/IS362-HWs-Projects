# Bicycle Trip: Daily Miles

This notebook uses a pandas Series to store cumulative odometer readings from a bicycle trip between NYC and Key West. Since an odometer shows the total distance traveled, the readings do not directly show the miles ridden on each individual day.

To find the daily mileage, I will keep the first day's reading because the trip began at zero. For each day after that, I will subtract the previous day's cumulative reading from the current day's reading.




## Importing pandas

I will import pandas using the abbreviation `pd`. Pandas is needed to create the Series that will hold the cumulative odometer readings.


```python
import pandas as pd
```

## Creating the cumulative mileage Series

The assignment provides cumulative odometer readings of 55 miles for day 1 and 120 miles for day 2. This means the rider traveled 55 miles on the first day and 65 miles on the second day.

Since the remaining readings are not provided, I assumed the bicycle trip lasted 25 days. Using an average of approximately 70 miles per day gives an estimated total of 1,750 miles. I created daily distances between 55 and 80 miles and used them to form the cumulative odometer readings.



```python

cumulative_miles = pd.Series(
    [55, 120, 192, 260, 335, 405, 485, 559, 628, 704,
     777, 848, 926, 993, 1068, 1140, 1210, 1289, 1357,
     1431, 1500, 1560, 1622, 1680, 1750],
    index=range(1, 26)
)

cumulative_miles
```

## Calculating the miles ridden each day

The first odometer reading is also the distance traveled on day 1 because the trip began at zero miles. For each remaining day, I will subtract the previous day's cumulative reading from the current day's reading. I will use a loop to repeat this calculation for all 25 days.


```python
daily_miles = [cumulative_miles.iloc[0]]

for day in range(1, len(cumulative_miles)):
    miles = cumulative_miles.iloc[day] - cumulative_miles.iloc[day - 1]
    daily_miles.append(miles)

daily_miles = pd.Series(daily_miles, index=range(1, 26))

daily_miles
```

## Results

To make the results easier to read, I placed the cumulative odometer readings and the calculated daily mileage next to each other in a pandas data frame. This makes it possible to compare the total distance at the end of each day with the distance ridden on that individual day.


```python
trip_data = pd.DataFrame({
    "Cumulative Miles": cumulative_miles,
    "Miles Ridden That Day": daily_miles
})

trip_data.index.name = "Day"

trip_data
```

 ## Checking the total mileage

To double check, the sum of the daily mileage should equal the last cumulative odometer reading. If both values equal 1,750 miles, the daily calculations are consistent with the original.


```python
daily_miles.sum()
```

## Conclusion

The daily mileage was calculated by finding the difference between consecutive odometer readings. For example, the rider traveled 65 miles on day 2 because the odometer increased from 55 miles to 120 miles. The final check also returned 1,750 miles, which matches the last cumulative reading. This means the individual daily distances account for the full trip.
