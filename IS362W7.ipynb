## Movie Ratings


I asked seven of my friends to rate six movies on a scale of 1 to 5. Not everyone had seen every movie, so some ratings are missing.

```python
import pandas as pd


ratings = {
    "Obsession": [5, None, None, 4.5, None, 4, 4],
    "Spider-Man: Brand New Day": [5, None, 4, 3.5, 4, 4, 4],
    "Marty Supreme": [4, None, None, None, 3.5, 3, 3],
    "The Super Mario Galaxy Movie": [3, None, 3, None, 4.5, None, 2],
    "Primetime": [3, 2, None, None, None, None, 2.5],
    "The Odyssey": [4, 3, 4, 4, None, None, None]
}

df = pd.DataFrame(ratings, index=[
    "Chris", "Ari", "Delina", "Alexandra",
    "Sydney", "Fatou", "Sere"
])

df
```

```python
df.to_csv("movie_ratings.csv")

```

## Saving the Data

I saved the ratings as a CSV file so I 
could load them back into pandas.

## Loading the CSV File


I loaded the CSV file into a DataFrame. The missing ratings appear as NaN.

```python
df = pd.read_csv("movie_ratings.csv", index_col=0)
df

```

## Average Ratings

I calculated the average rating for each person and each movie. Pandas ignores missing values when calculating the averages.


```python
print("Average rating by person:")
print(df.mean(axis=1))


print("\nAverage rating by movie:")
print(df.mean())
```

## Normalized Ratings

I used min-max normalization to put each person's ratings on a scale from 0 to 1. I subtracted their lowest rating and divided by the difference between their highest and lowest ratings. I did this separately for each person because everyone might use the 1 to 5 scale differently.



```python
normalized_df = df.copy()

for person in df.index:
    lowest = df.loc[person].min()
    highest = df.loc[person].max()

    if highest != lowest:
        normalized_df.loc[person] = (df.loc[person] - lowest) / (highest - lowest)
    else:
        normalized_df.loc[person] = 0

normalized_df
```

## Standardized Ratings

For the extra credit, I standardized each person's ratings using their average and standard deviation. This shows whether someone rated a movie higher or lower than their usual average. Unlike normalization, these values can also be negative.



```python
print("Average normalized rating by person:")
print(normalized_df.mean(axis=1))

print("\nAverage normalized rating by movie:")
print(normalized_df.mean())
```

## Standardized Ratings

For the extra credit, I standardized the ratings by subtracting each person's average and dividing by their standard deviation.


```python
standardized_df = df.copy()

for person in df.index:
    average = df.loc[person].mean()
    std = df.loc[person].std()

    if std != 0:
        standardized_df.loc[person] = (df.loc[person] - average) / std
    else:
        standardized_df.loc[person] = 0

standardized_df
```

## Average Standardized Ratings

I calculated the averages one more time using the standardized ratings.


```python
print("Average standardized rating by person:")
print(standardized_df.mean(axis=1))

print("\nAverage standardized rating by movie:")
print(standardized_df.mean())
```

## Conclusion

Normalizing the ratings makes it easier to compare people who rate movies differently. For example, someone who usually gives lower ratings can still show which movies they liked the most. One disadvantage is that normalization can make small differences in ratings seem bigger than they really are.

I also noticed that some movies had fewer ratings than others, which could affect how reliable their averages are. Overall, I think the original ratings are easier to understand, but normalization is useful for comparing preferences.
