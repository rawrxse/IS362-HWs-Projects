## Question 3: On February 12, 2013, which New York area airport had the windiest weather?

I will use the weather data to compare wind speeds at the New York area airports on this date.

```python
weather = pd.read_csv("weather.csv")
weather.head()
```

```python
feb12 = weather[
    (weather["year"] == 2013) &
    (weather["month"] == 2) &
    (weather["day"] == 12)
]

feb12
```

```python
feb12.sort_values("wind_speed", ascending=False).head(10)
```

There is a wind speed of 1048.36 mph for EWR, which is clearly an outlier and not a realistic wind speed. I will remove this value before comparing the airports.

```python
feb12_clean = feb12[feb12["wind_speed"] < 100]

feb12_clean.groupby("origin")["wind_speed"].max()
```

After removing the EWR outlier, I compared the highest wind speeds for each airport. LGA had the highest wind speed at about 23.02 mph.

Answer: LaGuardia Airport - LGA
