# Counting Repeated Clothing Items with Dictionaries 
I chose dictionary counting because I understand basic lists and dictionaries, but I struggle a little with updating a dictionary while looping through a list.


    

```python
clothing_items = ["dress", "jeans", "top", "dress", "skirt", "top", "dress", "jeans"]

print(clothing_items)

```

## Creating the List

I started by creating a list called clothing_items. Some of the clothing items repeat so that I can use a dictionary to keep track of how many times each one appears.


```python
item_counts = {}
print(item_counts)
```

## Counting the Clothing Items

Next, I will loop through the list and use the dictionary to keep track of how many times each clothing item appears.

```python
item_counts = {}

for item in clothing_items:
    item_counts[item] = item_counts.get(item, 0) + 1

print(item_counts)

```

### Understanding Get

`.get()` was the part I found most confusing. It checks whether an item is already a key in the dictionary. If the item is not there yet, `.get(item, 0)` gives it a starting value of 0. Then `+ 1` adds one to the count.

For example, the first time "dress" appears, its count starts at 0 and becomes 1. Each time "dress" appears again, its existing count increases by 1.


```python
print(item_counts.get("dress", 0))
print(item_counts.get("jacket", 0))
```

## Displaying the Final Counts

 Now I can loop through the completed dictionary to display each clothing item with its total count. This makes the results easier to read than printing the entire dictionary.
 
 

```python
for item, count in item_counts.items():
    print(item, ":", count)
    
```
