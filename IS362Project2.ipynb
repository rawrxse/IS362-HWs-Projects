# Project 2: Cleaning and Analyzing Untidy Data

For this project, I'll be working with three different datasets and using pandas to clean and organize them. I'll also analyze the data to see what patterns I can find.

# 
Dataset 1: Clothing Sales


For my first dataset, I used monthly sales for five clothing categories. I'll create a CSV file, load it into pandas, and reorganize it to make comparisons easier.

```python
import pandas as pd

clothing = {
    'Product': ['Jeans', 'Dresses', 'Sweaters', 'Skirts', 'Jackets'],
    'Jan': [120, 85, 175, 70, 200],
'Feb': [145, 110, 140, 90, 170],
'Mar': [160, 150, 95, 125, None]
    
}


df = pd.DataFrame(clothing)
df.to_csv('clothing_sales.csv', index=False)

df
```

```python
df = pd.read_csv('clothing_sales.csv')


df
```

# 
Checking for Missing Values

Before cleaning, I want to check if any sales values are missing.


```python
df.isnull().sum()

```

# Reshaping the Data

Right now, each month has its own column. I'll use melt() to put all the months into one column and their sales into another. This should make it easier to compare sales across categories.




```python
clothing_tidy = pd.melt(
    df,
    id_vars=['Product'],
    var_name='Month',
    value_name='Sales'
    
)

clothing_tidy
```

# Handling Missing Data

The March sales for Jackets are missing. Since I don't know the actual value, I'll remove that row so it doesn't affect my analysis.



```python
clothing_tidy = clothing_tidy.dropna()

clothing_tidy
```

# Total Sales by Category


Now that the data is organized, I'll add up the sales for each clothing category to see which one performed the best.


```python
category_sales = clothing_tidy.groupby('Product')['Sales'].sum()

category_sales
```

# Total Sales by Month

Next, I'll add up the sales for each month to see which month had the highest overall sales.



```python
monthly_sales = clothing_tidy.groupby('Month')['Sales'].sum()

monthly_sales
```

# Conclusion

Jeans had the highest total sales with 425 dollars. February had the highest overall sales at 655, with January very close at 650. March had the lowest total, but the missing jacket sales may have affected that result. 




# Dataset 2: NFL Quarterback Statistics (Dontae's)

This dataset shows passing yards and touchdowns for three quarterbacks from 2023 to 2025. The statistics are separated by year, so I'll reorganize the data to make it easier to compare their performance.


```python
nfl = {
    'Quarterback': ['Patrick Mahomes', 'Josh Allen', 'Lamar Jackson'],
    'Yards_2023': [4183, 4306, 3678],
    'Yards_2024': [3928, 3731, 4172],
    'Yards_2025': [4587, 3668, 4050],
    'TDs_2023': [27, 29, 24],
    'TDs_2024': [26, 28, 41],
    'TDs_2025': [30, 25, 34]
}

nfl_df = pd.DataFrame(nfl)
nfl_df.to_csv('nfl_stats.csv', index=False)

nfl_df
```

# Loading the Data

I'll load the CSV into pandas and check the data before
reorganizing it.

```python
nfl_df = pd.read_csv('nfl_stats.csv')

nfl_df
```

```python
nfl_df.isnull().sum()
```

# Reshaping the Data


The years are currently i
ncluded in the column names. I'll use melt() to put the statistics into a longer format, then separate the years from the types of statistics.

```python
nfl_tidy = nfl_df.melt(
    id_vars='Quarterback',
    var_name='Statistic_Year',
    value_name='Value'
)

nfl_tidy[['Statistic', 'Year']] = nfl_tidy['Statistic_Year'].str.split('_', expand=True)

nfl_tidy = nfl_tidy.drop(columns='Statistic_Year')

nfl_tidy
```

# Comparing Passing Yards

I'll compare the total passing 
yards for each quarterback across the three seasons.

```python
passing_yards = nfl_tidy[nfl_tidy['Statistic'] == 'Yards']

passing_yards.groupby('Quarterback')['Value'].sum()
```

# Comparing Touchdowns

Next, I'll add up the touchdowns
for each quarterback to see who had the most across the three seasons.

```python
touchdowns = nfl_tidy[nfl_tidy['Statistic'] == 'TDs']



touchdowns.groupby('Quarterback')['Value'].sum()
```

# Conclusion


Patrick Mahomes had the most passing yards across the three seasons with 12,698. Lamar Jackson had the most touchdowns with 99. Looking at the results, more passing yards doesn't necessarily mean more touchdowns


## Dataset 3: Book Sales (Alyssa's)

This dataset shows monthly sales for five books at a small bookstore. Each month has its own column, so I'll reorganize the data to compare sales by book, genre, and month.


```python
books = {
    'Book': ['The Last Summer', 'Midnight House', 'Lost in the Stars',
             'The Silent Case', 'Second Chances'],
    'Genre': ['Romance', 'Horror', 'Science Fiction',
              'Mystery', 'Romance'],
    'Jan': [45, 32, 27, 51, 39],
    'Feb': [52, 41, 35, 48, 44],
    'Mar': [61, 38, 46, 57, 50]
}

books_df = pd.DataFrame(books)
books_df.to_csv('book_sales.csv', index=False)

books_df
```

# Loading the Data

I'll read the CSV into pandas and check for missing values before cleaning it.


```python
books_df = pd.read_csv('book_sales.csv')

books_df
```

```python
books_df.isnull().sum()
```

# Reshaping the Data


The sales are spread across three monthly columns. I'll use melt() to put the months into one column and the sales into another.

```python
books_tidy = books_df.melt(
    id_vars=['Book', 'Genre'],
    var_name='Month',
    value_name='Sales'
)

books_tidy
```

# Total Sales by Book

I'll add up each book's sales to see which title sold the most over the three months.


```python
book_totals = books_tidy.groupby('Book')['Sales'].sum()

book_totals
```

# Sales by Genre

Next, I'll compare
the genres to see which one had the highest total sales.

```python
genre_totals = books_tidy.groupby('Genre')['Sales'].sum()

genre_totals
```

# Monthly Sales

I'll also compare the total sales for each month to see if sales increased or decreased over time.


```python
book_monthly = books_tidy.groupby('Month')['Sales'].sum()

book_monthly
```

# Conclusion

The Last Summer had the highest total sales with 158 books sold. Romance was the most popular genre overall, with 291 books sold. March had the highest sales at 252.
